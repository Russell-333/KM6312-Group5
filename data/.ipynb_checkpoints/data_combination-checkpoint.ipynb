```python
import pandas as pd

# 读取原始 CSV 文件
df = pd.read_csv("TiktokBan_comments_3&4.csv")

# 将 Date 列解析为 datetime 类型（处理 'Z' 和毫秒）
df['Date'] = pd.to_datetime(df['Date'], utc=True)

# 提取 2025 年 3 月的评论
march_df = df[(df['Date'].dt.year == 2025) & (df['Date'].dt.month == 3)]

# 提取 2025 年 4 月的评论
april_df = df[(df['Date'].dt.year == 2025) & (df['Date'].dt.month == 4)]

# 保存为新 CSV 文件
march_df.to_csv("March_comments.csv", index=False)
april_df.to_csv("April_comments.csv", index=False)

print("成功提取并保存 March_comments.csv 和 April_comments.csv")
```

```python
import pandas as pd

# 读取原始 CSV 文件
df = pd.read_csv("TiktokBan_comments_2.csv")

# 将 Date 列解析为 datetime 类型（处理 'Z' 和毫秒）
df['Date'] = pd.to_datetime(df['Date'], utc=True)

# 提取 2025 年 2 月的评论
February_df = df[(df['Date'].dt.year == 2025) & (df['Date'].dt.month == 2)]

# 保存为新 CSV 文件
February_df.to_csv("February_comments.csv", index=False)

print("成功提取并保存 February_comments.csv")
```

```python
import pandas as pd

# 读取三个 CSV 文件
df1 = pd.read_csv("TiktokBan_comments_1.1.csv")
df2 = pd.read_csv("TiktokBan_comments_1.2.csv")
df = pd.read_csv("TiktokBan_comments_1.3&12.1.csv")

# 将 Date 列解析为 datetime 类型（处理 'Z' 和毫秒）
df['Date'] = pd.to_datetime(df['Date'], utc=True)

# 提取 2025 年 1 月的评论
df3 = df[(df['Date'].dt.year == 2025) & (df['Date'].dt.month == 1)]

# 合并数据
January_df = pd.concat([df1, df2, df3], ignore_index=True)

# 去重
January_df.drop_duplicates(inplace=True)

# 保存到新文件
January_df.to_csv("January_comments.csv", index=False)

print("成功合并为 January_comments.csv")
```

```python
import pandas as pd

# 读取三个 CSV 文件
df = pd.read_csv("TiktokBan_comments_1.3&12.1.csv")
df2 = pd.read_csv("TiktokBan_comments_12.2.csv")
df3 = pd.read_csv("TiktokBan_comments_12.3.csv")
df4 = pd.read_csv("TiktokBan_comments_12.4.csv")

# 将 Date 列解析为 datetime 类型（处理 'Z' 和毫秒）
df['Date'] = pd.to_datetime(df['Date'], utc=True)

# 提取 2024 年 12 月的评论
df1 = df[(df['Date'].dt.year == 2024) & (df['Date'].dt.month == 12)]

# 合并数据
December_df = pd.concat([df1, df2, df3, df4], ignore_index=True)

# 去重
December_df.drop_duplicates(inplace=True)

# 保存到新文件
December_df.to_csv("December_comments.csv", index=False)

print("成功合并为 December_comments.csv")
```

```python
import pandas as pd

# 读取三个 CSV 文件
df_1 = pd.read_csv("December_comments.csv")
df_2 = pd.read_csv("January_comments.csv")
df_3 = pd.read_csv("February_comments.csv")
df_4 = pd.read_csv("March_comments.csv")
df_5 = pd.read_csv("April_comments.csv")

# 合并数据
Combined_df = pd.concat([df_1, df_2, df_3, df_4, df_5], ignore_index=True)

# 保存到新文件
Combined_df.to_csv("Combined_comments.csv", index=False)

print("成功合并为 Combined_comments.csv")
```
